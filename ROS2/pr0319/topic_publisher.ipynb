```python
import rclpy as rp
from geometry_msgs.msg import Twist

rp.init()
test_node = rp.create_node('pub_test')
```

```python
msg = Twist()
print(msg)
```

```python
msg.linear.x = 2.0
print(msg)
```

```python
pub = test_node.create_publisher(Twist, '/turtle1/cmd_vel', 10)
pub.publish(msg)
```

```python
msg.linear.x = 2.0
msg.angular.z = 2.0
pub.publish(msg)
print(msg)
```

```python
cnt = 0

def timer_callback():
    global cnt
    
    cnt += 1
    pub.publish(msg)

    if cnt > 3:
        raise Exception('Publisher Stop')
```

```python
timer_period = 2.0
timer = test_node.create_timer(timer_period, timer_callback)
```
